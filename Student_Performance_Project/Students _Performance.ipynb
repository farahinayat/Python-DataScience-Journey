### Data Cleaning

```python
df=df.drop_duplicates()
```

```python
df
```

```python
numeric_cols = ["Science", "English", "Maths", "History", "Attendance"]

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].mean()
)
```

### Filled the missing marks in a subject with average marks in that subject.

```python
df.isnull().sum().sum()
```

```python
df["Name"]=df["Name"].str.title()
```

```python
df
```

```python
df["AdmissionDate"]=pd.to_datetime(df["AdmissionDate"])
```

```python
df["Year"]=df["AdmissionDate"].dt.year
```

```python
df["Month"]=df["AdmissionDate"].dt.month
```

```python
df
```

### Feature Engineering

```python
df=df.copy()
```

```python
df["Total"]=(df["Science"]+df["English"]+df["History"]+df["Maths"])
```

```python
df["Average"]=df["Total"]/4
```

```python
df["Result"]=df["Average"].apply(lambda x:"Pass" if x>=60 else "Fail")
```

```python
df["Grade"]=df["Average"].apply(lambda x:"A" if x>=90 else "B" if x>=80 else "C" if x>=70 else "D" if x>=60 else "F")
```
