```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df1=pd.read_csv("Sales_Data_Cleaned.csv")
```

```python
df1
```

```python
sns.set_theme()
```

### Q1 — Countplot: Orders by Category

```python
sns.countplot(data=df1, x="Category")
plt.title("Orders by Category")
plt.show()
```

Electronics has the highest number of orders, while Office Equipment has the fewest.

### Q2 — Countplot with hue

Create a count plot showing:

Category → Order Status


```python
sns.countplot(data=df1,x="Category",hue="Product")
plt.title("Product Order")
plt.show()
```

Accessories category has more orders than other categories followed by the electronics.
Keyboard and mobile appears frequently.

### Q3 — Boxplot: Revenue by Category

```python
sns.boxplot(data=df1,x="Category",y="Revenue")
plt.title("Revenue by Category")
plt.show()
```

Which category has the highest median?
Which has the largest spread?
Are there potential outliers?

Electronic category has the highest median and largest spread of revenue data.

The Electronics revenue distribution appears positively/right-skewed, with some higher-value observations extending beyond the main distribution.

### Q4 — Violin Plot

```python
sns.violinplot(data=df1,x="Category",y="Revenue")
plt.title("Revenue by Category")
plt.show()
```

### Q5 — Product × Category

```python
sns.barplot(
    data=df1,
    x="Product",
    y="Revenue",
    hue="Category",
    estimator=np.sum
)
plt.xticks(rotation=45)
plt.title("Product_Revenue Comparison")
plt.show()
```

Mobile + Electronics are generating the highest revenue.

### Q8 — Monthly Revenue by Category ⭐

```python
sns.barplot(data=df1,x="MonthName",y="Revenue",hue="Category")
plt.title("Monthly Revenue by Category")
plt.show()
```

Electronics sales are more in the highest revenue month.

### Q10 — Correlation Heatmap

```python
numeric_df=df1.select_dtypes(include="number")

correlation=numeric_df.corr()
sns.heatmap(correlation,annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

Which variables have the strongest positive correlation? Price and Revenue (0.85)

Which variables have weak correlation? Quantity and Revenue (0.27)

Does Quantity have a strong correlation with Revenue? No, its value is 0.27 shows the weak relationship b/w Revenue and Quantity.

### Q11 — Analyst Challenge

Which product category has the highest revenue, and how does its order distribution compare with other categories?

```python
fig,axes=plt.subplots(1,2,figsize=(12,5))

sns.barplot(data=df1, x="Category",y="Revenue",estimator="sum", ax=axes[0])
axes[0].set_title("Revenue by Category")

sns.countplot(data=df1, x="Category", ax=axes[1])
axes[1].set_title("Order Distribution")

plt.xticks(rotation=45)
plt.show()
```

Highest revenue is generated by the electronics. Most of the orders are from this category followed by the accessories.

### Q12 — Mini Seaborn Dashboard

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.barplot(data=df1, x="Category",y="Revenue",estimator="sum", ax=axes[0,0])
axes[0,0].set_title("Revenue by Category")

sns.countplot(data=df1, x="Category", ax=axes[0,1])
axes[0,1].set_title("Order Distribution")

sns.boxplot(data=df1, x="Category",y="Revenue", ax=axes[1,0])
axes[1,0].set_title("Revenue Distribution")

sns.scatterplot(data=df1, x="Quantity", y="Revenue",hue="Category",ax=axes[1,1])
axes[1,1].set_title("Quantity vs Revenue")

plt.show()
```
