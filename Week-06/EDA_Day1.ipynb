### Part B — Data Cleaning

```python
df["Date"]=pd.to_datetime(df["Date"])
```

```python
df["OrderYear"]=df["Date"].dt.year
```

```python
df["OrderMonth"]=df["Date"].dt.month
```

```python
df["OrderDay"]=df["Date"].dt.day
```

```python
df
```

```python
df[df.isnull().any(axis=1)]
```

```python
df[df["Product"]=="Monitor"]
```

```python
df[df["Product"]=="Printer"]
```

```python
df.groupby("Product")["Price"].mean()
```

```python
df.groupby("Product")["Quantity"].mean()
```

```python
df["Price"]=df["Price"].fillna(26600.0)
```

The price charged for the printer in different month for Lahore city is samei.e 26600.0.

(Same price(29400.0)is charged for Multan across different orders) .So i put the same value in this for Lahore.

```python
df["Quantity"]=df["Quantity"].fillna(3.0) #fill with the median value
```

"The missing Quantity was filled with the median quantity because quantity is numerical and the median is less sensitive to extreme values."

```python
df[df.duplicated()]
```

```python
df=df.drop_duplicates()
```

```python
df.isnull().sum()
```
