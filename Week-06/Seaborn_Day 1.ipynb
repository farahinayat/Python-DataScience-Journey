```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("Sales_Data_Cleaned.csv")
```

```python
df
```

```python
sns.set_theme()
```

### Q1 — Bar Plot

Create Revenue by Product

```python
sns.barplot(x="Product",y="Revenue",data=df)
plt.title("Revenue by Product")
plt.xticks(rotation=45)
plt.show()
```

Laptop generated the highest average revenue followed by mobile.

### Q2 — Revenue by City

Create a Seaborn bar plot:

```python
sns.barplot(x="City",y="Revenue",data=df)
plt.title("Revenue by City")
plt.show()
```

Islamabad shows highest average revenue.

```python
city_revenue=df.groupby("City")["Revenue"].mean()
```

```python
plt.bar(city_revenue.index,city_revenue.values)
plt.title("Revenue by City")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.show()
```

This chart shows the Total Revenue by City that is max for Islamabad.

### Q3 — Product Revenue by Category

```python
sns.barplot(
    data=df,
    x="Product",
    y="Revenue",
    hue="Category"
)

plt.xticks(rotation=45)

plt.show()
```

hue parameter classify the x-variable into different groups and shows the respective values for each.

### Q4 — Revenue Distribution

```python
sns.histplot(data=df,x="Revenue")
plt.title("Revenue Distribution")
plt.show()
```

Above plot shows the rightward skewed distribution of revenue.

### Q5 — Quantity vs Revenue

```python
sns.scatterplot(data=df,x="Quantity",y="Revenue")
plt.title("Quantity vs Revenue")
plt.show()
```

The above scatter plot does not show any specific relation b/w Quantity and Revenue like previous one (Matplotlib chart)

### Q6 — Add Category with hue

```python
sns.scatterplot(data=df,x="Quantity",y="Revenue",hue="Category")
plt.title("Quantity vs Revenue")
plt.show()
```

Scatter plot with hue some how shows a linear relation between quantity and revenue for some products categories like office and accessories.

### Q7 — Box Plot

```python
sns.boxplot(data=df,x="Category",y="Revenue")
plt.xticks(rotation=45)
plt.show()
```

Investigate:
Which category has the highest median revenue?
Are there unusually high/low values?
Which category appears to have the greatest spread?

Electonic has highest median revenue.

As the median for this category lies near the bottom and the data is skewed to the right,it has some outliers.

This category also has greatest spread of data as compared to other product categories.

### Q8 — Compare Products by Category

```python
sns.boxplot(data=df,x="Product",y="Revenue")
plt.title("Product Revenue")
plt.xticks(rotation=45)
plt.show()
```

Laptop shows the greatest variation in order revenue.

### 9 — Seaborn vs Matplotlib

1. Seaborn is comparatively easy to use as compared to the Matplotlib.

2. It categorize the x variable by default and plot the data while Matpotlib needs a code to group data first before plotting.

3. It also has a useful feature of box plot that gives valuable insights about the data like outliers and data spread.

4. Seaborn by default plot the aggregate values for the given variable while Matplotlib needs to calculate the aggregate first.


### Mini Challenge

Create one visualization that answers:

Which product category appears to generate the highest revenue, and how is revenue distributed within that category?

```python
fig, axs = plt.subplots(ncols=2, figsize=(12, 5))

sns.barplot(data=df, x="Product", y="Revenue", estimator=np.sum,ax=axs[0])
axs[0].set_title("Product Revenue")
plt.xticks(rotation=45)

sns.histplot(data=df, x="Revenue", ax=axs[1])
axs[1].set_title("Revenue Distribution by Product")

plt.tight_layout()
plt.xticks(rotation=45)
plt.show()
```
