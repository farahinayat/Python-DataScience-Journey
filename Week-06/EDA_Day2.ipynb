### Part 1 — Prepare Your Data

```python
import pandas as pd
df=pd.read_csv("Sales_Data_Cleaned")
```

```python
df
```

```python
df.dtypes
```

```python
df["Revenue"]=df["Quantity"]*df["Price"]
```

```python
df["Date"]=pd.to_datetime(df["Date"])
```

```python
df["MonthName"]=df["Date"].dt.month_name()
```

### EDA Day 2 Challenge

```python

product_summary = df.groupby("Product").agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Average_Price=("Price", "mean")
)

product_summary["Revenue_per_unit"] = (
    product_summary["Total_Revenue"] /
    product_summary["Total_Quantity"]
)

product_summary
```

Laptop has higher revenue per unit followed by mobiles and monitors. The months(April,June,Jan) with higher demand for these products generated the highest revenues.
