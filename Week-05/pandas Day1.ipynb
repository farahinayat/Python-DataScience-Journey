### Create a Series

```python
import pandas as pd
marks=pd.Series([80,75,90,88,65])
print(marks)
```

```python
print(type(marks))
print(marks.size)
print(marks.shape)
print(marks.dtype)
```

### Create a DataFrame

```python
student={
    "Name":["Ali","Sara","Ahmed","Ayesha"],
    "Age":[22,21,23,20],
    "Marks":[80,75,90,88]
}
df=pd.DataFrame(student)
print(df)
```

### View Data
