```python
import pandas as pd

Students=pd.DataFrame({
    "Student ID":[101,102,103,104],
    "Name":["Ali", "Sara", "Ahmed", "Fatima"]
})

Marks=pd.DataFrame({
    "Student ID":[101,102,103,104],
     "Math": [85, 92, 78, 90],
    "English": [80, 88, 75, 91],
})

Students
Marks
    
```

### Merge DataFrames

```python
df=pd.merge(Students,Marks, on="Student ID")
df
```

```python
df1=pd.merge(Students, Marks, on="Student ID", how="inner")
df1
```

```python
df2=pd.merge(Students, Marks, on="Student ID", how="outer")
df2
```

```python
df3=pd.merge(Students, Marks, on="Student ID", how="right")
df3
```

```python
df2=pd.merge(Students, Marks, on="Student ID", how="left")
df2
```

### Updates DataFrames

```python
Students = pd.DataFrame({
    "Student ID":[101,102,103,104,105],
    "Name":["Ali","Sara","Ahmed","Fatima","Bilal"]
})

Marks = pd.DataFrame({
    "Student ID":[101,102,103,106],
    "Math":[85,92,78,95],
    "English":[80,88,75,90]
})

Students
Marks
```

```python
df1=pd.merge(Students,Marks,on="Student ID",how="inner" )
df1
```

### inner keeps only the matching ids

```python
df2=pd.merge(Students,Marks,on="Student ID",how="outer" )
df2
```

### outer keeps all the ids from both data frames

```python
df3=pd.merge(Students,Marks,on="Student ID",how="right" )
df3
```

### right join keeps all ids from marks

```python
df4=pd.merge(Students,Marks,on="Student ID",how="left" )
df4
```

### left join keeps all the ids from students

### Concatenate

```python
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara"],
    "Math": [85, 90]
})

df2 = pd.DataFrame({
    "Name": ["Ahmed", "Fatima"],
    "Math": [78, 92]
})

pd.concat([df1,df2],ignore_index=True)
```

### Merge with Different Column Names

```python
student_info = pd.DataFrame({
    "ID": [101, 102, 103],
    "Name": ["Ali", "Sara", "Ahmed"]
})

scores = pd.DataFrame({
    "StudentID": [101, 102, 103],
    "Marks": [88, 91, 79]
})
```

```python
pd.merge(student_info, scores,left_on="ID", right_on="StudentID")
```
