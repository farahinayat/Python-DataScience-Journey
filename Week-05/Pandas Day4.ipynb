```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Bilal", "Fatima"],
    "Age": [20, np.nan, 22, 21, np.nan, 23],
    "City": ["Lahore", "Islamabad", np.nan, "Karachi", "Lahore", "Multan"],
    "Math": [85, 92, np.nan, 89, 68, 95],
    "English": [78, np.nan, 81, 91, 74, 93]
}

df = pd.DataFrame(data)

df
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.notnull()
```

```python
df.dropna()
```

```python
df.dropna(axis=1)
```

```python
df.fillna(0)
```

```python
df["Age"]=df["Age"].fillna(df["Age"].mean())
```

```python
df
```

```python
df["Math"]=df["Math"].fillna(df["Math"].mean())
```

```python
df["City"]=df["City"].fillna("Unknown")
```

```python
df["English"]=df["English"].fillna(df["English"].mean())
```

```python
df
```

```python
df["City"]=df["City"].replace("Lahore","Lahore City")
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
df["City"]=df["City"].replace("Karachi","Karachi City")
```

```python
df
```

```python
df.to_csv("cleaned_students.csv", index=False)
```
