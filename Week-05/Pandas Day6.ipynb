### Pivot Tables

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali","Sara","Ahmed","Fatima","Ali","Sara"],
    "City": ["Lahore","Karachi","Lahore","Islamabad","Lahore","Karachi"],
    "Subject": ["Math","Math","English","English","Science","Science"],
    "Marks": [85,92,78,88,90,95]
})

df
```

```python
pd.pivot_table(df,values="Marks",index="City")
```

```python
pd.pivot_table(df,values="Marks",index="City",columns="Subject")
```

```python
pd.pivot_table(df,values="Marks",index="Name",columns=["City","Subject"])
```

```python
pd.pivot_table(df,values="Marks",index="City",aggfunc="sum")
```

```python
df.pivot_table(values="Marks",index="City",aggfunc="sum")
```

```python
pd.pivot_table(df,values="Marks",index="City",aggfunc="count")
```

```python
pd.pivot_table(df,values="Marks",index="Subject",aggfunc="mean")
```

```python
pd.pivot_table(df,values="Marks",index="City",aggfunc="max")
```

```python
pd.pivot_table(df,values="Marks",index="Subject",aggfunc="min")
```

```python
pd.pivot_table(df,values="Name",index="City",aggfunc="count")
```

### Crosstab

```python
pd.crosstab(df["City"],df["Subject"])
```

```python
pd.crosstab(df["City"],df["Name"])
```

### Date & Time

```python
dates=pd.DataFrame({
    "OrderDate":["2026-05-15","2026-05-17","2026-05-20","2026-05-25","2026-06-15"]
})

dates
```

```python
dates["OrderDate"]=pd.to_datetime(dates["OrderDate"])
```

```python
dates["Year"]=dates["OrderDate"].dt.year
```

```python
dates["Month"]=dates["OrderDate"].dt.month
```

```python
dates["Day"]=dates["OrderDate"].dt.day
```

```python
dates["DayName"]=dates["OrderDate"].dt.day_name()
```

```python
dates
```
