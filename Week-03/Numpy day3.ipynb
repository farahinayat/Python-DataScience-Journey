```python
#Real Mini Project

sales = np.array([
    [120,150,130,140],
    [200,210,190,220],
    [160,170,180,175]
])
print(sales)
```

```python
#1 total sales
print(f"Total sales = {sales.sum()}.")
```

```python
#2 average sales
print(f"Average sales = {sales.mean()}.")
```

```python
#3 max sales
print(f"Max sale = {sales.max()}.")
```

```python
#4 min sales
print(f"Min sale = {sales.min()}.")
```

```python
#5 total sales of each salesperson (axis=1)
sales.sum(axis=1)
```

```python
#6 Average sales of each month (axis=0)
sales.mean(axis=0)
```

```python
#7 Which salesperson sold the most?
sales.sum(axis=1) #second sales person has max sales 
```

```python
#8 Which month had the highest average sales?
sales.mean(axis=0)  #4th month has highest aeverage sales.
```

```python
# challenge2
marks = np.array([
    [80,75,90,85],
    [70,88,78,82],
    [95,91,89,94]
]) 
print(marks)
```

```python
#Which student scored the highest total?
marks.sum(axis=1) #last one has the highest total marks.
```

```python
#Which subject has the highest average?
marks.mean(axis=0) #last one subject has the highest average.
```

```python
#Which student scored the highest in Subject 2?
marks.max(axis=0) #highest marks in subject 2 is 91(in last row) for last one student.
```
