```python
#Mini project
marks = np.array([
    [80,75,90,85],
    [70,88,78,82],
    [95,91,89,93],
    [60,72,68,70],
    [85,80,84,88]
])

print(marks)
print(marks.shape)
print(marks[2,:]) #marks of third student
print(marks[:,1]) #marks in second subject
```

```python
marks.max()
```

```python
marks.min()
```

```python
marks.mean()
```

```python
marks.mean(axis=0)
```

```python
marks.mean(axis=1)
```

```python
marks.sum(axis=1) #third student has highest total marks
```
