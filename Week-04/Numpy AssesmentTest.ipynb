Section E – Practical

```python
marks = np.array([
[65,72,81],
[55,90,78],
[88,69,91]
])
print(marks)
```

```python
#Total marks of all students.
marks.sum()
print(f"Total marks of all students are {marks.sum()}")
```

```python
marks.mean()
print(f"Average marks of all students are {marks.mean()}")
```

```python
marks.max()
print(f"Highest marks are {marks.max()}")
```

```python
marks.min()
print(f"Lowest marks are {marks.min()}")
```

```python
print(f" Marks of second student are {marks[1,:]}")
```

```python
print(f" Marks in third subject are {marks[:,2]}")
```

```python
print(marks+5)
```

```python
print(marks.T)
```

```python
print(marks.flatten())
```

```python
print(marks.shape)
```
