Section B – Coding Basics

```python
#Print numbers from 1 to 100.
for i in range(1,101):
    print(i)
```

```python
#Print only odd numbers between 1 and 100.
for i in range(1,101):
    if i%2!=0:
        print(i)
```

```python
#Find the sum of numbers from 1 to 100.
total=0
for i in range(1,101):
    total=total+i
    i+=1

print(total)
```

```python

n=int(input("Enter a num"))
fact=1
for i in range(1,n+1):
    fact=fact*i
    i+=1
print(fact)
```

```python
#Print the multiplication table of any number entered by the user.
n=int(input("Enter a num"))
for i in range(1,11):
    print(n*i)
```

Section C – Pattern Printing

```python
for i in range(1,6):
    print(i*"*")
```

```python
for i in range(1,6):
    for j in range(i):
        print(i, end="")
    print()
```

```python
for i in range(1,6):
    for j in range(i):
        print(j+1, end="")
    print()
```

Section D – Functions

```python
def square(n):
    print(n**2)

square(5)
```

```python
def cube(n):
    print(n**3)

cube(5)
```

```python
def evenodd(n):
    if n%2==0:
        print("n is even")
    else:
        print("n is odd")

evenodd(5)
        
```

```python
def check(a,b,c):
    if a>b and a>c:
        print("a is largest")
    elif b>a and b>c:
        print("b is largest")
    elif c>a and c>b:
        print("c is largest")
check(9,8,4)
    
```

```python
def reverse(s):
    print(s[::-1])

reverse("farah")
```

Section E – Lists

```python
marks = [45,78,67,89,23,90,56]
print(marks)
```

```python
print(min(marks))
```

```python
print(max(marks))
```

```python
print(sum(marks)/len(marks))
```

```python
for n in marks:
    if n>60:
        print(n)
```

```python
marks = [45,78,67,89,23,90,56]
marks.sort(reverse=True)
print(marks)
```

```python
marks.append(100)
print(marks)
```

```python
marks.remove(min(marks))
print(marks)
```
