## Relation Between GDP Growth and Unemployment

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=economic_data,
    x="GDP_Growth",
    y="Unemployment_Rate",
    hue="Country",
    s=80
)

plt.title("GDP Growth vs Unemployment Rate: 2000–2025")
plt.xlabel("GDP Growth (%)")
plt.ylabel("Unemployment Rate (%)")
plt.grid(True, alpha=0.3)
plt.show()
```

## Correlation Findings

The correlation analysis shows a very weak relationship between GDP
growth and unemployment across the five selected countries.

The correlation coefficient for the combined sample is -0.051,
indicating almost no linear relationship between the two variables.

For Pakistan, the correlation coefficient is -0.259, indicating a
weak negative relationship. This suggests that higher GDP growth
periods tend to coincide with somewhat lower unemployment rates,
although the relationship is not strong.

These results represent correlation rather than causation. Therefore,
the analysis does not establish that changes in GDP growth directly
cause changes in unemployment.

## Country-by-Country Economic Performance Comparison

This section compares the economic performance of Pakistan, India,
Bangladesh, Sri Lanka, and Nepal using average GDP growth, average
unemployment, extreme values, and the correlation between GDP growth
and unemployment.

The comparison helps identify differences in economic growth and labor
market outcomes across the selected countries.

```python
country_performance=(
    economic_data
    .groupby("Country")
    .agg(
        Avg_GDP_Growth=("GDP_Growth","mean"),
        Avg_Unemployment=("Unemployment_Rate","mean"),
        Max_GDP_Growth=("GDP_Growth","max"),
        Min_GDP_Growth=("GDP_Growth","min"),
        Max_Unemployment=("Unemployment_Rate","max"),
        Min_Unemployment=("Unemployment_Rate","min")
    )
    .round(2)
)

country_performance
```

```python
country_performance["GDP_Growth_Rank"]=(
    country_performance["Avg_GDP_Growth"]
    .rank(ascending=False)
    .astype(int)
)
```

```python
country_performance.sort_values("GDP_Growth_Rank")
```

```python
country_performance["Unemployment_Rank"]=(
    country_performance["Avg_Unemployment"]
    .rank(ascending=True)
    .astype(int)
)
    
    
```

```python
country_performance.sort_values("Unemployment_Rank")
```

## Average GDP Growth by Country: 2000–2025

```python
plt.figure(figsize=(9,5))

sns.barplot(
    data=country_performance.sort_values(
        "Avg_GDP_Growth",
        ascending=False
    ),
    x=country_performance.sort_values(
        "Avg_GDP_Growth",
        ascending=False
    ).index,
    y="Avg_GDP_Growth"
)

plt.title("Average GDP Growth by Country: 2000–2025")
plt.xlabel("Country")
plt.ylabel("Average GDP Growth (%)")
plt.xticks(rotation=45)
plt.show()
```

India recorded the highest average GDP growth at 6.25%, followed by Bangladesh at 5.87%. Pakistan recorded the lowest average GDP growth among the five countries at 3.96%.

## Average Unemployment by Country: 2000–2025

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=country_performance.sort_values(
        "Avg_Unemployment",
        ascending=False
    ),
    x=country_performance.sort_values(
        "Avg_Unemployment",
        ascending=False
    ).index,
    y="Avg_Unemployment"
)

plt.title("Average Unemployment by Country: 2000–2025")
plt.xlabel("Country")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

Nepal recorded the highest average unemployment rate at 10.75%, while Pakistan recorded the lowest at 2.58%.

```python
country_correlations = (
    economic_data
    .groupby("Country")
    .apply(
        lambda x: x["GDP_Growth"].corr(x["Unemployment_Rate"]),
        include_groups=False
    )
    .rename("GDP_Unemployment_Correlation")
    .to_frame()
)

country_correlations.round(3)
```

```python
country_performance=country_performance.join(country_correlations)

country_performance.round(2)
```

```python
comparison=country_performance.reset_index()

comparison[["Country","Avg_GDP_Growth","Avg_Unemployment"]]
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=comparison,
    x="Avg_GDP_Growth",
    y="Avg_Unemployment",
    hue="Country",
    s=150
)

plt.title("Average GDP Growth vs Average Unemployment")
plt.xlabel("Average GDP Growth (%)")
plt.ylabel("Average Unemployment Rate (%)")
plt.grid(True, alpha=0.3)
plt.show()
```

## Country-by-Country Economic Performance

The country comparison shows substantial differences in economic
performance across the five selected South Asian economies.

India recorded the highest average GDP growth at 6.25%, followed by
Bangladesh at 5.87%. Pakistan recorded the lowest average GDP growth
at 3.96%, closely followed by Sri Lanka at 3.99%.

In terms of average unemployment, Pakistan recorded the lowest rate
at 2.58%, while Nepal recorded the highest at 10.75%.

The relationship between GDP growth and unemployment varies across
countries. Nepal shows the strongest negative correlation (-0.45),
while Pakistan shows a weak negative correlation (-0.26). Sri Lanka
(0.03) and Bangladesh (0.11) show almost no linear relationship.

These results suggest that higher GDP growth does not automatically
translate into lower unemployment across all countries. Labor-market
conditions can differ substantially between economies.

The correlations indicate association rather than causation, so the
results should not be interpreted as evidence that GDP growth directly
causes changes in unemployment.

## Conclusion

This project analyzed the relationship between economic growth and
unemployment across Pakistan, India, Bangladesh, Nepal, and Sri Lanka
from 2000 to 2025.

The analysis found considerable differences in economic performance
across the five countries. India recorded the highest average GDP
growth at 6.25%, followed by Bangladesh at 5.87%. Pakistan recorded
the lowest average GDP growth at 3.96%.

Despite its relatively low GDP growth, Pakistan recorded the lowest
average unemployment rate among the five countries at 2.58%. Nepal
recorded the highest average unemployment rate at 10.75%.

For Pakistan, the highest GDP growth occurred in 2004 at 7.83%, while
the lowest occurred in 2020 at -1.27%. Pakistan also experienced
negative GDP growth in 2020 and 2023.

The correlation analysis showed a weak negative relationship between
GDP growth and unemployment in Pakistan, with a correlation coefficient
of -0.26. Across all five countries combined, the correlation was
approximately -0.05, indicating almost no linear relationship.

These findings suggest that GDP growth and unemployment do not have a
strong or consistent linear relationship across the selected countries
during the study period. Differences in labor markets, economic
structure, demographics, and other economic factors may influence
unemployment independently of GDP growth.

The results should be interpreted as descriptive associations rather
than evidence of causation. Correlation alone cannot establish that
changes in GDP growth directly cause changes in unemployment.
